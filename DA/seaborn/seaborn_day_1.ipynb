```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
d={
    "Data":[
        "2025-01-01",
        "2025-01-02",
        "2025-01-03",
        "2025-01-04",
        "2025-01-05"
    ],
    "Sales":[5000,6500,4200,7000,5500],
    "Sales-2":[5000,650,200,7000,550]}
df=pd.DataFrame(d)
print(df)
```

```python
sns.get_dataset_names()
```

```python
ds=sns.load_dataset('diamonds')
print(ds)
```

```python
sns.lineplot(data=ds,x='color',y='price',hue='cut',marker='*')
plt.title('line plot')
plt.show()
```

```python
sns.barplot(data=ds,x='cut',y='price',hue='color')
plt.title('bar graph')
plt.show()
```

```python
sns.scatterplot(data=ds,x='carat',y='price',hue='cut')
plt.title('scatter-plot')
plt.show()
```

```python
sns.boxplot(data=ds,x='cut',y='price',hue='color')
plt.title('box-plot')
plt.show()
```

```python
sns.histplot(data=ds,x='price')
plt.title('Histogram')
plt.show()
```

```python
sns.violinplot(data=ds,x='cut',y='price')
plt.title("vionlin polt")
plt.xlabel("Cut")
plt.ylabel("Price")
plt.show()
```
