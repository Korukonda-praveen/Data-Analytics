```python
# Product prices
Prices=pd.Series([500,750,1200,900,1500],
index=['P101','P102','P103','P104','P105'])
print("--Before and changes---")
print(Prices)
print('\n--- Display on the element with index P103 ---')
print(Prices['P103'])

print('\n--- change the elemet 750 to 800 ---')
Prices['P102']=800
print(Prices)


print('\n--- After Drop ---')
Prices=Prices.drop('P104')
print(Prices)
print("\n--- After Addition ---")
Prices['P106']=2000
print(Prices)
```

```python
# daily temperature
temp= pd.Series([31,32,34,33,35,36,34],
index=['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
print('--- Before any changes ---')
print(temp)

print('\n--- Display the temperature from Monday to Wednesday ---')
print(temp[:3])

print("\n--- Change Thursday's temperature to 32 ---")
temp['Thursday']=32
print(temp)

print("\n--- Delete Saturday's temperature ---")
temp=temp.drop('Saturday')
print(temp)

print('\n--- Add Monday of yhe next week wih temperature ---')

temp['next_Monday']=30
print(temp)

```

```python
# Bank Account Balances
Balances=pd.Series([25000,45000,32000,10000,60000],
index=['A101','A102','A103','A104','A105'])
print('--- Before any changes ---')
print(Balances)

print('\n--- Display the account balance of A103 ---')
print(Balances['A103'])

print('\n--- Update A104 balance to 25000 ---')
Balances['A104']=25000
print(Balances)

print('\n--- Delete Account A101 ---')
Balances=Balances.drop('A101')
print(Balances)

print('\n--- Add Account A106 with balance 40000 ---')
Balances['A106']=40000
print(Balances)
```

What is Dataframe:
a Dataframe is a two-dimensional labeled data structure in pandas
DataFrame contains rows and columns

 we can create dataframe using "pd.DataFrame()"



```python
# dataframe creation
# using list
data=[
    ["ravi",22,30000],
    ["priya",24,40000],
    ["Arun",21,28000]
    ]

df=pd.DataFrame(data,columns=['Name','Age','Salary'])
print(df)
```

```python
# using tuple creating a dataframe
data=(
    ("jaya",22,30000),
    ("tony",24,40000),
    ("dhoom",21,28000)
)

df=pd.DataFrame(data,columns=['Name','Age','Salary'])
print(df)
```

```python
# using dict creating a dataframe
data={
    'Name':['ravi','priya','Arun'],
    'Age':[22,24,21],
    'Salary':[30000,40000,28000]
}

df=pd.DataFrame(data)
print(df)
```

```python
# create a dataframe sales and profit every month first 6 months of the year
sales=[
      ['jan',100,2000],
      ['feb',200,3000],
      ['mar',300,4000],
      ['apr',400,5000],
      ['may',500,6000],
      ['jun',600,7000]
]
df=pd.DataFrame(sales,columns=['Month','Sales','Profit'])
print(df)
```

```python
# create a dataframe using set
my_set={100,200,300,400}

df=pd.DataFrame(list(my_set),columns=['values'])
print(df)
```

```python
# creating single series into dataframe
s=pd.Series([1,2,3,4],name='Numbers')
df=s.to_frame()
print(df)
```
