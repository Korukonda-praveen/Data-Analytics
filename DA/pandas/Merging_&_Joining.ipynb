```python
# Merging & Joining
import pandas as pd
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Salary': [100, 200, 300]
})

```

```python
merged = pd.merge(
    df1,
    df2,
    on='ID',
    how='inner')
print(merged)
```

```python
# outer
m1 = pd.merge(
    df1,
    df2,
    on='ID',
    how='outer'
)
print(m1)
```

```python
# right
m2 = pd.merge(
    df1,
    df2,
    on='ID',
    how='right'
)
print(m2)
```

```python
# left
m3 = pd.merge(
    df1,
    df2,
    on='ID',
    how='left'
)
print(m3)
```

```python
# pd.concat() for rows
combined = pd.concat(
    [df1, df2],
    axis=0,
    ignore_index=True
)
print(combined)
```

```python
# for columns
c1 = pd.concat([df1, df2],axis=1,ignore_index=True)
print(c1)
```

```python
# day 13
# apply()

import pandas as pd
dept_data= {
      "Department":['IT','HR','Finance','Sales'],
      "Salary":[35000,42000,48000,55000],
      "Location":['Hyderabad','Bangalore','Chennai','Delhi']
}
dept_df=pd.DataFrame(dept_data)
dept_df
```

```python
dept_df['Bonus']=dept_df['Salary'].apply(lambda x:x*0.10)
dept_df
```

```python
city_code={
    "Hyderabad":'HYD',
    "Bangalore":'BLR',
    "Chennai":'CHE',
    "Delhi":'DEL'
}
dept_df['city_code']=dept_df['Location'].map(city_code)
dept_df
```
