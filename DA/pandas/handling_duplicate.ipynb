handling duplicate values

df.duplicated() for find the duplicate row

df.duplicated(keep=False)To find all rows which are invloved

By default keep will be in True

```python
import pandas as pd
data = {
    "EmpID":[101,102,103,104,105,106,107,108,109,110,110],
    "Name":['Ravi','PRIYA','Arun','sneha','Kiran','ANJALI','Vikram',
  'meena','RAHUL',
      'Sneha','Sneha'],
    "City":["Hyderabad","hyderabad","CHENNAI","Chennai",
    "Banglore","BANGLORE",
        'Mumbai','mumbai','Delhi','Hyderabad','Hyderabad'],
    "Salary":[30000,
    4000,3500,4500,32000,5000,38000,42000,36000,45000,45000]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.duplicated()
```

```python
print(df[df.duplicated()])
```

```python
d=df[df.duplicated(keep=False)] # to print both original and duplicate value
print(d)
```

```python
count=df.duplicated().sum()
print(count)
```

```python
duplicate_rows = df[df.duplicated()]
total = duplicate_rows['Salary'].sum()
print(total)
```

```python
duplicate_rows = df[df.duplicated()]
total = duplicate_rows[['Name','Salary']].sum()
print(total)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]
print(duplicate_rows)
```

```python
df['EmpID'].duplicated()
```

```python
f=df['Name'].duplicated() #storing
print(f)
```

```python
# To drop the duplicates we use df=df.drop_duplcates()
df.drop_duplicates()

print(df)
```

```python
print(df)
```

```python
df.drop_duplicates(keep='last')
```

```python
df.drop_duplicates(keep='first')
```

```python
# remove all occurrences of duplicates
df.drop_duplicates(keep=False)
```

```python

data = {
   "TransactionID": [1001, 1002, 1003, 1004, 1005,
                      1006, 1007, 1008, 1009, 1010, 1010, 1011],
   "Customer": ["Ravi", "Priya", "Arun", "Sneha", "Kiran",
                "Anjali", "Vikram", "Meena", "Rahul", "Sneha",
                "Sneha", "Kiran"],
   "Product": ["Laptop", "Mobile", "Tablet", "Headphones", "Mouse",
               "Keyboard", "Monitor", "Laptop", "Mobile", "Headphones",
               "Headphones", "Mouse"],
   "Amount": [55000, 25000, 30000, 5000, 1500,
              2500, 18000, 55000, 22000, 5000,
              5000, 1500]}
df_sales = pd.DataFrame(data)
print(df_sales)


```

```python
# 1
df_sales[df_sales.duplicated()]
```

```python
# 2
count=df_sales.duplicated().sum()
print(count)
```

```python
# 3
duplicate_sales=df_sales[df_sales.duplicated(keep=False)]
print(duplicate_sales)
```

```python
# 4
ds=df_sales[df_sales.duplicated(keep=False)]
total=ds['Amount'].sum()
print(total)
```

```python
# 5
ds=df_sales[df_sales.duplicated(keep=False)]
t=ds['TransactionID'].count()
print(t)
```

```python
# 6
df_sales[df_sales['Customer'].duplicated(keep=False)]
```

```python
# 7
df_sales.drop_duplicates(keep='first')
```

```python
# 8
df_sales['TransactionID'].drop_duplicates()
```

```python
# 9
df_sales[['Customer','Product']].drop_duplicates()
```

```python
# 10
df_sales.drop_duplicates(['TransactionID'],keep='first',inplace=True)
print(df_sales)
```
