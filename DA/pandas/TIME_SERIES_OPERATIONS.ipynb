```python
import pandas as pd
data={
    'Date':[
        '2025-01-01',
        '2025-01-02',
        '2025-01-03',
        '2025-01-04',
        '2025-01-05'
    ],
    "Sales":[5000,6500,4200,7000,5500]
}
df=pd.DataFrame(data)
print(df)
df.dtypes
```

```python
# pd.to_datatime() is used converts values into pandas datatime format
df['Date']=pd.to_datetime(df['Date'])
df.dtypes
```

```python
# pd.data_range is used to show specifying and end data with no of periods
dates=pd.date_range(
    start='2025-01-01',
    periods=10
)
print(dates)
```

```python
# Generating Data with Frequency
'''
freq='D' is used of Day
freq='H' is used for hourly
freq='min' is used for minute
freq='MS' is used month starting
freq='ME' is used month ending
freq='YS' is used year starting
freq='YE' is used year ending
'''
dates = pd.date_range(start="2025-01-01",periods=10,freq="D")
print(dates)
```

```python
dates=pd.date_range(start='2025-01-01',periods=10,freq='H')
print(dates)
```

```python
dates=pd.date_range(start='2025-01-01',periods=10,freq='min')
print(dates)
```

```python
dates=pd.date_range(start='2025-01-01',periods=10,freq='MS')
print(dates)
```

```python
dates=pd.date_range(start='2025-01-01',periods=10,freq='ME')
print(dates)
```

```python
dates=pd.date_range(start='2025-01-01',periods=10,freq='YS')
print(dates)
```

```python
dates=pd.date_range(start='2025-01-01',periods=10,freq='YE')
print(dates)
```

```python
df = pd.DataFrame({
    "DateTime": pd.to_datetime([
        "2025-01-01 08:30:15",
        "2025-01-01 10:45:20",
        "2025-01-01 14:20:30",
        "2025-01-01 18:10:45"
    ]),
    "Sales": [2000, 3500, 5000, 4200]
})
df["DateTime"].dt.hour
```
