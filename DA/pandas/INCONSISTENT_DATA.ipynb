```python
# copy() method
data={
    'Name':['ravi','priya','arun','sneha'],
    'Age':[22,24,21,25],
    'Salary':[30000,40000,28000,50000]
}
df=pd.DataFrame(data)
print(df)
```

```python
d1=df.copy()
print(d1)
```

```python
d1['Name']=d1['Name'].str.strip().str.title()
print('original data')
print(df)
print('copied data')
print(d1)
```

```python
df_deep=df.copy(deep=True)
df_deep.loc[0,'Salary']=50000
print('original data')
print(df)
print('deep copy')
print(df_deep)
```

```python
df_shallow=df.copy(deep=False)
df_shallow.loc[2,'Age']=20
print('original data')
print(df)
print('shallow copy')
print(df_shallow)
```
