```python
import plotly.express as px

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

fig = px.line(
    x=x,
    y=y
)

fig.show()
```

```python
# create a line using plotly by taking the values of
# gold price drop every month

x=['jan','feb','mar','apr','may','jun','jul','aug','sep','oct','nov','dec']
y=[15000,16000,14000,17000,18000,12000,9000,8000,5500,6000,4000,6000]

f=px.line(
    x=x,
    y=y,
    title='gold price drop every month using line chart',
    labels={"x": "Month", "y": "Price Drop (₹)"}
)
f.show()
```

```python
f=px.bar(
    x=x,
    y=y,
    title='gold price drop every month bar chart',
    labels={"x": "Month", "y": "Price Drop (₹)"}
)
f.show()
```

```python
f=px.scatter(
    x=x,
    y=y,
    title='gold price drop every month scatter',
    labels={"x": "Month", "y": "Price Drop (₹)"}
)
f.show()
```

```python
# create bar,line,scatter based on the data given

food_items = ["rice pack", "wheat flour", "cooking oil", "sugar", "pulses"]
prices = [550, 320, 180, 95, 240]
fi = px.line(
    x=food_items,
    y=prices,
    title="Food items and prices using line chart",
    labels={"x": "Food items", "y": "Prices"},
)
fi.show()
```

```python
fi = px.bar(
    x=food_items,
    y=prices,
    title="Food items and prices using bar chart",
    labels={"x": "Food items", "y": "Prices"},
)
fi.show()
```

```python
fi = px.scatter(
    x=food_items,
    y=prices,
    title="Food items and prices using scatter",
    labels={"x": "Food items", "y": "Prices"},
)
fi.show()
```

```python
fi=px.pie(
    names=food_items,
    values=prices,
    title='Food items and prices using pie chart'
)
fi.show()
```

```python
fi=px.pie(
    names=food_items,
    values=prices,
    hole=0.6,
    title='Food items and prices using donut chart'
)
fi.show()
```
