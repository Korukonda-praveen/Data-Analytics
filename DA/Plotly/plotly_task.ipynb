#Retail Sales Data Analysis and Interactive Visualization Using Python and Plotly

```python
import pandas as pd

df=pd.read_csv("/content/retail_sales_dataset.csv")
print(df)
```

```python
import plotly.express as px
```

```python
# Bar chart
fig_bar=px.bar(
    df,
    x='Product Category',
    y='Total Amount',
    color='Product Category',
    title='Total sales by Product Category',
     labels={
        'Product Category': 'Category',
        'Total Amount': 'Sales Amount'}
)
fig_bar.show()
```

```python
# line chart
fig_line=px.line(
    df.head(100),
    x='Transaction ID',
    y='Total Amount',
    markers=True,
    title='Sales by Transacion',
    labels={'Transaction ID':'Transaction Number', 'Total Amount':'Sales Amount'}
)
fig_line.show()
```

```python
# scatter
fig_scatter=px.scatter(
    df,
    x='Age',
    y='Total Amount',
    color='Product Category',
    title='Customer age Vs Sales Amount',
    labels={'Age':'Customer Age',
            'Total Amount':'Sales Amount'}

)
fig_scatter.show()
```

```python
# pie chart
fig_pie=px.pie(
    df,
    names='Product Category',
    values='Total Amount',
    title='Sales Distribution by Product Category',
    labels={'Product Category':'Category','Total amount':'Sales Amount'}
)
fig_pie.show()
```

```python
# donut
fig_donut=px.pie(
    df,
    names='Product Category',
    values='Quantity',
    hole=0.5,
    title='Quantity Sold by Product Category',
   labels={'Product Category':'Category',
            'Quantity':'Total Quantity'}

)
fig_donut.show()
```

```python
# box plot
fig_box=px.box(
    df,
    x='Product Category',
    y='Total Amount',
    color='Product Category',
    title='Sales Distribution by Product Category',
    labels={'Product Category':'Category','Total Amount':'Sales Amount'}
)
fig_box.show()
```

```python
# Violin Plot
fig_violin=px.violin(
    df,
    x='Gender',
    y='Total Amount',
    color='Gender',
    box=True,
    points='all',
    title='Customer Spending Distribution by Gender',
    labels={'Gender':'Customer Gender','Total Amount':'Sales Amount'}
)
fig_violin.show()
```
