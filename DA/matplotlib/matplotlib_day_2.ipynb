```python
# Stacked Bar Chart
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({
    'Products':['Laptop','Mobile','Tablet','Monitor'],
    'Sales':[500,700,400,600],
    'Profit':[100,150,80,120]
})
df.plot(
    x='Products',
    y=['Sales','Profit'],
    kind='bar',
    width=0.5
)
plt.title('Sales and Product')
plt.xlabel('Product')
plt.ylabel('Amount')
plt.legend()
plt.show()
```

```python
# Stacked column Chart
products=['a','b','c','d','e']
sales=[200,300,400,250,350]
profit=[100,150,200,120,150]
plt.bar(products, sales, label='sales')
plt.bar(products, profit, bottom=sales, label='profit')

plt.title('Sales and Profit')
plt.xlabel('Product')
plt.ylabel('Amount')

plt.legend()

plt.show()
```

```python
#  Area Chart
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = [100, 120, 150, 140, 180, 200]
plt.plot(months, monthly_sales,color='black')
plt.fill_between(months, monthly_sales,color='orange')
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# pie chart
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Product Sales Distribution")
plt.show()
```
