```python
import matplotlib.pyplot as plt
import numpy as np
products=['Laptop','Mobile','Tablet','Headphones','Watch']
sales=[120,180,90,150,110]
profit=[30,50,20,40,25]
plt.plot(products,sales,label='Sales',color='red')
plt.title('Sales and Profit Analysis')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
plt.bar(products,sales,label='Profit',color='blue')
plt.title('Products and Sales Analysis')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
plt.barh(products,profit,label='Profit',color='green')
plt.title('Products and Profit Analysis')
plt.ylabel('Products')
plt.xlabel('Profit')
plt.legend()
plt.show()
```

```python
plt.plot(products,sales,label='Sales',marker='o',color='black')
plt.plot(products,profit,label='Profit',marker='^',color='red')
plt.title('multiple line chart/Sales VS Profit')
plt.xlabel('Products')
plt.ylabel('Amount')
plt.legend()
plt.show()
```

```python
# dataset
months=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales=[25000,30000,28000,35000,40000,38000,45000,42000,50000,48000,55000,60000]
profit=[5000,7000,6000,8000,10000,9000,12000,11000,14000,13000,15000,18000]
expenses=[20000,23000,22000,27000,30000,29000,33000,31000,36000,35000,40000,42000]
```

```python
# 1
plt.plot(months, sales, label='sales',marker='*')
plt.title('Total sales across 12 months')
plt.xlabel('months')
plt.ylabel('sales')
plt.legend()
plt.show()
```

```python
# 2
plt.bar(months,sales,label='sales',color='blue')
plt.title('jan to dec revenue')
plt.xlabel('months')
plt.ylabel('sales')
plt.legend()
plt.show()
```

```python
# 3
plt.plot(months,sales,label='Sales',marker='o',color='blue')
plt.plot(months,profit,label='Profit',marker='^',color='orange')
plt.title('Sales VS Profit')
plt.xlabel('sales')
plt.ylabel('profit')
plt.legend()
plt.show()
```

```python
# 4
plt.bar(months,expenses,label='expenses',color='green')
plt.title('expenses over months')
plt.xlabel('months')
plt.ylabel('expenses')
plt.legend()
plt.show()
```

```python
# 6
plt.plot(months,sales,label='sales',marker='*',color='pink')
plt.title('sales over year')
plt.xlabel('months')
plt.ylabel('sales')
plt.legend()
plt.show()
```

```python
plt.plot(months, sales, label='sales', marker='*', color='blue')
plt.plot(months, expenses, label='expenses', marker='^', color='black')  # Fixed 'maker' to 'marker'
plt.plot(months, profit, label='profit', marker='o', color='pink')     # Fixed 'maker' and added quotes to label

plt.title('sales/expenses/profit')
plt.xlabel('months')
plt.legend()
plt.show()
```
