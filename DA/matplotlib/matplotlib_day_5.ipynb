```python
# Contour plot
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(-1,10)
y=np.arange(-1,11)
x,y=np.meshgrid(x,y)
# z=x**2+y**2
z=np.tan(x)+np.cos(y)
plt.contour(x,y,z)
plt.title('Contour Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# Violin Plot
marks=[60, 65, 70, 72, 75, 80, 85]
plt.violinplot(marks)

plt.title("Distribution of Student Marks")
plt.ylabel("Marks")

plt.show()
```

```python
# stem plot
values = [10, 15, 20, 18, 25, 30, 28]

plt.stem(values)

plt.title("Stem Plot")

plt.show()
```

```python
# heatmap
data = [
    [80, 70, 90],
    [60, 85, 75],
    [90, 95, 80]
]
plt.imshow(data)

plt.title("Performance Heatmap")

plt.xlabel("Subjects")
plt.ylabel("Students")

plt.colorbar()

plt.show()
```
