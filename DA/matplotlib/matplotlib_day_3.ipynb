```python
# Scatter plot
import matplotlib.pyplot as plt
advertising = [10, 20, 30, 40, 50, 60, 70, 80]
sales_amount = [25, 35, 40, 50, 55, 65, 72, 85]
plt.scatter(advertising, sales_amount)
plt.title("Advertising vs Sales")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.show()
```

```python
# donut chart
plt.pie(
    advertising,
    labels=sales_amount,
    autopct="%1.1f%%",
    wedgeprops={"width":0.6}
)

plt.title("Advertising vs Sales")

plt.show()
```

```python
# 1
dep = ['HR', 'Sales', 'IT', 'Marketing', 'Finance']
count = [20, 45, 30, 25, 15]

plt.pie(
    count,
    labels=dep,
    autopct="%1.1f%%",
)
plt.title('Dep and count')

plt.show()
```

```python
# 2
act=['Sleeping','Studying','Playing','Eating','Other Activities']
hours=[8,5,3,2,6]

plt.pie(
    hours,
    labels=act,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.6}
)
plt.text(0,0,
    '24 hours',
    ha="center",
    va="center"

)
plt.title('24 hours and activities')
plt.show()
```
