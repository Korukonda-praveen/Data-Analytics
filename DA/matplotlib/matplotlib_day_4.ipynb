```python
# box plot
import matplotlib.pyplot as plt
import numpy as np

marks=[-50,45,52,44,55,75,65,56,67,87,23,45,29,90,20,200]
plt.boxplot(marks)
plt.title('Student Marks')
plt.ylabel('Marks')
plt.show()
```

```python
python_marks=[60,65,70,72,75,80,85,78,200]
sql_marks=[55,60,65,70,72,78,82,300]
powerbi_marks=[50,60,68,72,75,80,90,400]
```

```python
plt.boxplot(
    [python_marks, sql_marks, powerbi_marks],
    labels=['python','sql','power bi']
)
plt.title('Student Performance')
plt.ylabel('Marks')
plt.show()
```

```python
# histogram
marks=[35,45,52,44,55,75,65,56,67,87,23,45,29,90,20]
plt.hist(marks,bins=10,edgecolor='black')
plt.xlabel('marks')
plt.ylabel('frequency')
plt.title('histo')
plt.show()
```

```python
# create a histogram for a bookstore sales of 10 books


book_rates = [125, 155, 175, 185, 158, 185, 135, 115, 195, 199]

plt.hist(book_rates, bins=10, edgecolor='black',color='red')
plt.xlabel('book rates')
plt.ylabel('frequency')
plt.title('Bookstore Sales Distribution')

plt.show()
```

```python
plt.boxplot(book_rates)
plt.title('Bookstore Sales Distribution')
plt.ylabel('book rates')
plt.legend()
plt.show()
```

```python
pen_sales=[50,70,90,120,150,67,89,170,185,190]
plt.hist(pen_sales,bins=10,edgecolor='black',color='orange')
plt.xlabel('pen sales')
plt.ylabel('frequency')
plt.title('Pen Sales Distribution')

plt.show()
```

```python
book_rates = [125, 155, 175, 185, 158, 185, 135, 115, 195, 199]
pen_sales=[50,70,90,120,150,67,89,170,185,190]
plt.scatter(book_rates,pen_sales)
plt.title('books vs pens')
plt.xlabel('books')
plt.ylabel('pens')
plt.show()
```
