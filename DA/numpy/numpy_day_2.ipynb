```python
# for zeroes np.zeros
import numpy as np
d=np.zeros(8)
print(d)
```

```python
# for ones np.ones and number in the brackets
d1=np.ones(4)
print(d1)
```

```python
# for same number we need to use full
s=np.full(5,7)
print(s)
```

```python
# linspace is used for numbers get random numbers between  range with points decimals
# the decimals point will be same for the all the points
arr=np.linspace(1,11,5)
print(arr)
```

```python
# arange used for show the sepecific data from one range to another range
arr2=np.arange(1,11)
print(arr2)
```

```python
# reshape to shape of the array will make sure the rows
# and columns is equal to the number of elements
print(arr2.reshape(1,10))
print(arr2.reshape(2,5))
```

```python
# for the Identity Matrix
print(np.eye(5))
```
