```python
#NumPy Random Package
#it generating random numbers.
# 1D
import numpy as np  #0 to 1
arr=np.random.random(5)
print(arr)
```

```python
# 2D
arr2=np.random.random((2,3))
print(arr2)
```

```python
# 3D
arr3=np.random.random((2,2,3))
print(arr3)
```

```python
# uniform() Method
# np.random.uniform(low, high, size)
arr4=np.random.uniform(10,50,5)
print(arr4)
```

```python
# randn() Method
arr5 = np.random.randn(5)
print(arr5)
print(arr5.mean())
print(arr5.std())
```

```python
# choice() Method is used select random from the array
arr6=np.array([10,20,30,40,50])
result=np.random.choice(arr6)
print(result)
```

```python
# for selecting multiple values
arr7=np.array([10,20,30,40,50])
result=np.random.choice(arr7,3)
print(result)
```

```python
# choice() with Probability
arr8 = [10, 20, 30]
result = np.random.choice(
    arr8,
    size=5,
    p=[0.2, 0.5, 0.3]
)

print(result)
```

```python
# seed() Method it is used to want  same random values every time

arr9=np.random.random(5)
np.random.seed(2)
print(arr9)
```

```python
# shuffle() Method
arr = np.array([10, 20, 30, 40, 50])
np.random.shuffle(arr)
print(arr)
```

```python
# permutation() Method
arr = np.array([10, 20, 30, 40, 50])
result = np.random.permutation(arr)
print("Original:", arr)
print("Permutation:", result)

```
