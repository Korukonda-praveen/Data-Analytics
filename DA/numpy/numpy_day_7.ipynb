```python
import numpy as np
```

```python
# Splitting Arrays – split()
# Split 1D Array
arr = np.array([10, 20, 30, 40, 50, 60])
result = np.split(arr, 3)
print(result)
```

```python
# Split 1D Array at Specific Positions
arr=np.array([10,20,30,40,50,60])
result=np.split(arr,[2,4])
r1=np.split(arr,[3,4])
r2=np.split(arr,[1,3])
r3=np.split(arr,[2,3])
r4=np.split(arr,[1,2,3])

print(result)
print('-'*50)
print(r1)
print('-'*50)
print(r2)
print('-'*50)
print(r3)
print('-'*50)
print(r4)
```

```python
# Split 2D Array Horizontally
arr=np.array([
    [10,20,30,40],
    [50,60,70,80]
])
result=np.split(arr, 2, axis=1)
print(result)
```

```python
result=np.split(arr, 2, axis=0)
print(result)
```

```python
# Split 3D Array
a=np.arange(1,13).reshape(2,2,3)
b=np.arange(1,13).reshape(3,2,2)
c=np.arange(1,13).reshape(2,3,2)

print(a)
print('-'*50)
print(b)
print('-'*50)
print(c)
```

```python
r=np.split(a,2,axis=0)
print(r)
print('-'*50)
r=np.split(a,2,axis=1)
print(r)
print('-'*50)
```
