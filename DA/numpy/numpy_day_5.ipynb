```python
import numpy as np
# np.log10() calculates the base-10 logarithm.
arr = np.array([10, 100, 1000])
print(np.log10(arr))
```

```python
# np.log2() calculates the base-2 logarithm.
arr1=np.array([2,4,8,16])
print(np.log2(arr1))
```

```python
# np.log() calculates the natural logarithm, using base e.
# e ≈ 2.71828
arr = np.array([1, 2, 10])
print(np.log(arr))
```

```python
# np.abs() returns the absolute value.
arr = np.array([-10, -5, 0, 5, 10])
print(np.abs(arr))
```

```python
# NumPy provides np.sort() for sorting arrays.
arr = np.array([50, 10, 40, 20, 30])
result = np.sort(arr)
print(result)
```

```python
# Sorting String Array
names = np.array([
    "Ravi",
    "Arun",
    "Priya",
    "Kiran"])
result = np.sort(names)
print(result)
```

```python
#  Sorting in Decreasing Order Using Slicing
arr = np.array([50, 10, 40, 20, 30])
result = np.sort(arr)[::-1]
print(result)
```

```python
#  Data Type Conversion
# Integer → Float
arr = np.array([10, 20, 30])
new_arr = arr.astype(float)
print(new_arr)
print(new_arr.dtype)
```

```python
# Float → Integer
arr = np.array([10.5, 20.8, 30.9])
new_arr = arr.astype(int)
print(new_arr)
print(new_arr.dtype)
```

```python
# Integer → String
arr = np.array([10, 20, 30])
new_arr = arr.astype(str)
print(new_arr)
print(new_arr.dtype)
```

```python
# String -> Integer
arr = np.array(['10', '20', '30'])
new_arr = arr.astype(int)
print(new_arr)
print(new_arr.dtype)
```

```python
# Slicing a 2D Array
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print(arr[0:2])
```

```python
# Modification
arr = np.array([10, 20, 30, 40, 50])
arr[2] = 100
print(arr)
```
