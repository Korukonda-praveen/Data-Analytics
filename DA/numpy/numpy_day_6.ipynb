```python
# Transpose the Array
import numpy as np
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]])
print(arr)
```

```python
result = arr.transpose()
print(result)
```

```python
# ravel() Method
# ravel() converts an array into a one-dimensional array.
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
result = arr.ravel()
print(result)
```

```python
arr=np.array([10,20,30,40])
new_arr=np.insert(arr,2,100)
print(arr)
print(new_arr)
```

```python
# Insert Multiple Values
arr = np.array([10, 20, 30, 40])
new_arr = np.insert(arr, 2, [100, 200])
print(new_arr)
```

```python
# Deletion
arr = np.array([10, 20, 30, 40, 50])
new_arr = np.delete(arr, 2)
print(new_arr)
```

```python
# Delete Multiple Elements
arr = np.array([10, 20, 30, 40, 50])
new_arr = np.delete(arr, [1, 3])
print(new_arr)
```

```python
# 2D Array Horizontal Concatenation

arr1=np.array([
    [10,20],
    [30,40]
])
arr2=np.array([
    [50,60],
    [70,80]

])
result=np.concatenate((arr1,arr2),axis=1)
result1=np.concatenate((arr1,arr2),axis=0)
print(result)
print('-'*25)
print(result1)
```

```python
# np.stack()
# np.stack() combines arrays by creating a new axis.
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])
result = np.stack((arr1, arr2), axis=0)
print(result)

```

```python
result = np.stack((arr1, arr2), axis=1)
print(result)
```

```python
# 2D Arrays Stacking

arr1 = np.array([
    [1, 2],
    [3, 4]
])
arr2 = np.array([
    [5, 6],
    [7, 8]
])
result = np.stack((arr1, arr2), axis=0)
result1 = np.stack((arr1, arr2), axis=1)
print(result)
print('-'*25)
print(result1)
```

```python
# 3d
arr1=np.arange(1,11,2)
arr2=np.arange(11,21,2)
print(arr1)
print(arr2)
```

```python
p=arr1.reshape(1,5,1)
p2=arr2.reshape(1,5,1)
print(p)
print(p2)
print('-'*25)
result=np.stack((p,p2),axis=0)
result1=np.stack((p,p2),axis=1)
print(result)
print('-'*25)
print(result1)
```
