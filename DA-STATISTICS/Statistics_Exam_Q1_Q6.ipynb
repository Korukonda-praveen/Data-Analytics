```python
# 4
pop=np.array([2,4,4,6])
mean=np.mean(pop)
print(f'(a)Population Mean:{mean}')
# (b)deviations adn squared deviations
dev=pop-mean
sq_dev=dev**2
print(f'(b)Deviations from mean:{dev.tolist()}')
print(f'    Squared deviations:{sq_dev.tolist()}')

# (c) Calculate the population variance and population standard deviation.
pop_var=np.var(pop)
pop_std=np.std(pop)
print(f'(c) Population Variance:{pop_var:.1f}')
print(f' Population Standard Deviation:{pop_std:.2f}')

# (d)Sample variance
sample_var=np.var(pop)
print(f'(d)sample Variance:{sample_var:.1f}')
```

```python
# 5
data=np.array([4,6,7,8,9,10,12,30])
q1=np.percentile(data,25)
q3=np.percentile(data,75)
print(f'(a) Q1:{q1:.2f} | Q3:{q3:.2f}')


# (b) Interquratile range(IQR)
iqr=q3-q1
print(f'(b) IQR: {iqr:.2f}')


#(c) lower and upper fences using 1.5 *iqr
lower_fence=q1-1.5*iqr
upper_fence=q3+1.5*iqr
print(f'(c) Lower Fence:{lower_fence:.2f} | Upper Fence:{upper_fence:.2f}')


# (d) observations flagged by fences

flagged=data[(data<lower_fence)|(data>upper_fence)]
print(f'(d) Flagged Observations (outliers):{flagged.tolist()}')

# (e) Bocplot whisker
lower_whisker = np.min(data[data >= lower_fence])
upper_whisker = np.max(data[data <= upper_fence])
print(f"(e) Lower Whisker Endpoint: {lower_whisker}")
print(f"    Upper Whisker Endpoint: {upper_whisker}")
```
