```python
# 5
According to the Central Limit Theorem (CLT):

When the sample size increases, the distribution of sample means becomes approximately normal. The sample mean distribution is centred around the population mean. The standard error decreases as the sample size increases. Does individual spending become normal?

No. Individual customer spending remains right-skewed.

Conclusion: CLT makes the sampling distribution of means approximately normal, not the individual data.
```

```python
#6
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

mean = 10
sd = 10
sample_sizes = [2, 10, 50, 200]

for n in sample_sizes:
  means = []
  for _ in range(5000):
    sample = np.random.exponential(scale=mean, size=n)
    means.append(np.mean(sample))

  centre = np.mean(means)
  sim_se = np.std(means, ddof=1)
  theo_se = sd / np.sqrt(n)

  print("n =", n)
  print("Centre:", round(centre, 2))
  print("Simulated SE:", round(sim_se, 2))
  print("Theoretical SE:", round(theo_se, 2))
  print()

  plt.hist(means, bins=30,edgecolor='black')
  plt.title("n = " + str(n))
  plt.show()
```
