```python
# Standard Deviation and
# Standard Error (Variation of an estimate)-->How much sample average vary..
import math
import statistics
scores=[55,60,65,70,75]
sample_size=len(scores)
sample_mean=statistics.mean(scores)
sample_std=statistics.stdev(scores) #do this mathematically at home
estimated_se=sample_std/math.sqrt(sample_size)  #we are getting to the possible sample
print(f'Sample size is : {sample_size}')
print(f'Sample mean is : {sample_mean}')
print('Sample Std is  :',round(sample_std,3))
print('Estimated error:',round(estimated_se,2)) # estimated variation of sample means
```

```python
# Generate independent waiting times from an exponential population model
import numpy as np
import matplotlib.pyplot as plt
# Create a range -->random range
rng=np.random.default_rng(42)
print(rng)
sample_sizes=[1,5,30,100]
repetitions=5000
fig,axes=plt.subplots(2,2,figsize=(10,7))
for index,sample_size in enumerate(sample_sizes):
  sample_means=[]
  for repetition in range(repetitions):
    sample=rng.exponential(scale=10,size=sample_size)
    sample_means.append(np.mean(sample))
  simulated_se=np.std(sample_means,ddof=1)
  therotical_se=10/np.sqrt(sample_size)
  print(sample_size,
        np.mean(sample_means),
        simulated_se,
        therotical_se)
  ax=axes[index//2,index%2]
  #print(ax)
  ax.hist(sample_means,bins=40,color='mediumpurple',edgecolor='black')
  ax.axvline(10,color='gold',linestyle="--")
  ax.set_title(f'Sample means: n={sample_size}')
  ax.set_xlabel('Mean Waiting time (minutes)')
  ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
#graph is showing out sample_size,avg of simulated means,simulated Se,theretical SE
```
