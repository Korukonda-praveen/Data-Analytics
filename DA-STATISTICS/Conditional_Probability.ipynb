```python
#Reusable function-->Compare different priors
# As in above case intial spam percentage is 20%,after obser the word offer-->60%in the
def spam_probability(prior,offer_given_spam,offer_given_not_spam):
  """Spam probability Reusable"""
  not_spam=1-prior
  spam_and_offer=prior*offer_given_spam
  not_spam_and_offer=(not_spam*offer_given_not_spam)

  probability_offer=(spam_and_offer+not_spam_and_offer) #law of probability
  if probability_offer==0:
    raise ValueError('Canoot condition on a Zero probability event')
  posterior=spam_and_offer/probability_offer
  return posterior
```

```python
priors=[0.2,0.01,0.05,0.75,]
for prior in priors:
  result=spam_probability(prior,
                            offer_given_spam=0.6,
                          offer_given_not_spam=0.1)
  print("prior:",round(prior*100,2),
        "% | Posterior:",round(result*100,2),
        "%")
```
