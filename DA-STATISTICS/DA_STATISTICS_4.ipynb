```python
# Now we are going link this to rolling a die and getting an even  number
# Simulate a die
import random
trails=10000
random.seed(42)
even_count=0
for trial in range(trails):
  result=random.randint(1,6)
  if result%2==0:
    even_count+=1
estimated_probability=even_count/trails
print(f'Even Outcomes: {even_count}')
print(f'Estimated Probability: {round(estimated_probability,3)}')
print('Theoretical Probability:',0.5)
```
